```python
import pandas as pd
```

```python
df = pd.read_csv("../predictions/example_T2_primary_v1_cells_moderator.csv")
```

```python
df
```

```python
import json

# 1. JSON containing all unique conditions
conditions = (
    df["condition"]
    .dropna()
    .drop_duplicates()
    .sort_values()
    .tolist()
)

with open("../qstn_data/conditions.json", "w", encoding="utf-8") as f:
    json.dump(conditions, f, indent=2, ensure_ascii=False)


# 2. JSON containing each moderator and its possible levels
moderators = (
    df[["moderator", "moderator_level"]]
    .dropna()
    .drop_duplicates()
    .groupby("moderator")["moderator_level"]
    .apply(lambda values: sorted(values.tolist()))
    .to_dict()
)

with open("../qstn_data/moderators.json", "w", encoding="utf-8") as f:
    json.dump(moderators, f, indent=2, ensure_ascii=False)
```
